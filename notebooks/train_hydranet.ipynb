# HydraNet Multi-Task Training (skeleton)

AuraView K-Perception — 점수 **AI활용 · 학습 5점** 증빙용.

**백본:** YOLOv8n (Ultralytics)  
**Heads:** vehicle, VRU, signal, lane, sign  
**데이터:** AIHub 도로장애물 + Roboflow K-LISA + 자체 라벨링

````python
from ultralytics import YOLO
import os

DATA_YAML = 'datasets/auraview/hydranet.yaml'
model = YOLO('yolov8n.pt')
results = model.train(data=DATA_YAML, epochs=30, imgsz=640, batch=16, name='hydranet_v0')
````

````python
# 평가
metrics = model.val()
print({'map50': metrics.box.map50, 'map': metrics.box.map})
````
